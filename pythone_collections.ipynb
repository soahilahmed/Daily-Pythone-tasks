nums = [5, 2, 9, 1, 5]

 -Add number 10 at end

 -Insert 7 at index 2

 -Remove value 5 (first occurrence)

 -Sort the list

 -Count how many times 5 appears

```python
nums = [5, 2, 9, 1, 5]
```

```python
# Re-initialize nums to ensure a clean state
nums = [5, 2, 9, 1, 5]

# Add number 10 at end
nums.append(10)
print(f"After appending 10: {nums}")

# Insert 7 at index 2
nums.insert(2, 7)
print(f"After inserting 7 at index 2: {nums}")

# Remove value 5 (first occurrence)
nums.remove(5)
print(f"After removing the first 5: {nums}")

# Sort the list
nums.sort()
print(f"After sorting: {nums}")

# Count how many times 5 appears
count_of_5 = nums.count(5)
print(f"Count of 5s in the list: {count_of_5}")
```
